# **Pengantar Web Mining**

## Apa itu Web Mining?

Web Mining atau penambangan web merupakan proses menemukan dan mengekstrak informasi dari internet menggunakan berbagai teknik penambangan data. Informasi yang diperoleh dapat digunakan untuk menemukan pola dan pengetahuan yang berguna.

Web memiliki jumlah data yang sangat besar dan kompleks. Data pada web dapat berupa HTML, XML, dan teks. Selain itu, web bersifat dinamis dan tidak terbatas pada satu bidang tertentu sehingga proses penambangan data pada web memiliki berbagai tantangan.

## **Jenis Web Mining**

1. Web Content Mining

Web Content Mining merupakan proses mengekstrak informasi yang berguna dari dokumen-dokumen web. Fokusnya adalah pada konten halaman web, seperti teks, gambar, audio, video, serta data terstruktur seperti tabel dan daftar.

Salah satu bagian dari Web Content Mining adalah Text Mining, yang dapat digunakan untuk melakukan klasifikasi dokumen, pengelompokan dokumen, ekstraksi informasi, topic modelling, analisis sentimen, dan ekstraksi kata kunci.

2. Web Usage Mining

Web Usage Mining merupakan proses ekstraksi informasi dari data yang dihasilkan melalui kunjungan halaman web dan transaksi. Data tersebut dapat berasal dari server log, referer log, agent log, cookie, maupun data clickstream.

3. Web Structure Mining

Web Structure Mining merupakan proses penemuan dan interpretasi pola pada struktur tautan (hyperlink) di internet serta hubungan sosial antaraktor yang berinteraksi di web. Struktur web dapat direpresentasikan sebagai grafik yang terdiri dari node dan edge.

## Proses Web Mining

1. Gathering and Exploration

Yaitu mengumpulkan data web melalui crawling dokumen/data atau Web API, kemudian melakukan eksplorasi untuk memahami data.

2. Preprocessing and Transformation

Mengubah data ke dalam bentuk yang sesuai dengan metode data mining, termasuk seleksi fitur, reduksi dimensi, dan transformasi teks menjadi vektor atau embedding.

3. Actual Data Mining

Menggunakan metode data mining terhadap data yang telah diproses untuk menghasilkan model atau pola, kemudian mengevaluasi hasilnya dan melakukan perbaikan apabila diperlukan.
